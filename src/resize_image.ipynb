```python
# LIBRARY
import cv2
import matplotlib.pyplot as plt
```

```python
# Lihat aset
logo1 = cv2.imread('../data/raw/small_payload.png', 0)
logo2 = cv2.imread('../data/raw/medium_payload.png', 0)
logo3 = cv2.imread('../data/raw/high_payload.png', 0)

plt.imshow(logo1, cmap = 'gray')
plt.title("Small Payload") # Disamakan dengan nama file
plt.axis('off')
plt.show()

plt.imshow(logo2, cmap = 'gray') # Typo log2 diperbaiki menjadi logo2
plt.title("Medium Payload")      # Disamakan dengan nama file
plt.axis('off')
plt.show()

plt.imshow(logo3, cmap = 'gray')
plt.title("High Payload")
plt.axis('off')
plt.show()
```

```python
# RESIZE TO SMALL PAYLOAD
# resize
logo1_small = cv2.resize(logo1, (128,128))

# thresholding ubah piksel jadi hitam putih
_, logo1_small_bw = cv2.threshold(logo1_small, 127, 255, cv2.THRESH_BINARY) # # _, untuk mengabaikan nilai return pertama (threshold value)

#  show
plt.imshow(logo1_small_bw, cmap='gray')
plt.title("Small Payload")
plt.axis('off') # menghilangkan garis sumbu x dan y agar rapi outputny
plt.show()
print("Dimensi piksel logo1: ", logo1_small_bw.shape)
```

```python
# RESIZE TO MEDIUM PAYLOAD
# resize
logo2_medium = cv2.resize(logo2, (512, 512))

# thresholding ubah piksel jadi hitam putih
_, logo2_medium_bw = cv2.threshold(logo2_medium, 127, 255, cv2.THRESH_BINARY) # cv2.THRESH_BINARY cek tiap piksel jika nilai > 127 jadi 255. jika <= 127 jadi 0

# show
plt.imshow(logo2_medium_bw, cmap='gray')
plt.title('Medium Payload')
plt.axis('off')
plt.show()
print("Dimensi piksel logo1: ", logo2_medium_bw.shape)
```

```python
# RESIZE TO HIGH PAYLOAD
# resize
logo3_high = cv2.resize(logo3, (1024, 1024))

# thresholding ubah piksel jadi hitam putih
_, logo3_high_bw = cv2.threshold(logo3_high, 127, 255, cv2.THRESH_BINARY) # threshold manual 127 belum tentu optimal bisa pakai Otsu's method yang menghitung threshold otomatis berdasarkan histogram

# show
plt.imshow(logo3_high_bw, cmap='gray')
plt.title('High Payload')
plt.axis('off')
plt.show()
print("Dimensi piksel logo3: ", logo3_high_bw.shape)
```
